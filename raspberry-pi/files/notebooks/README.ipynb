This tool should feel familiar to Google Colab, but there are some differences. To run code in a cell:

1. Click the cell, your cursor should be in the cell with the code
2. Click the run button at the top of the page:  
   ![Run button](./assets/readme/jupyterlab-run-button.png)
   You can also hit "Shift" and then "Enter" to do the same thing

You can tell if a cell's code has run by looking at the square brackets (`[ ]`) to the left of the cell:

![Cell run status indicator](./assets/readme/jupyterlab-cell-run-indicator.png)

- If the cell has not run this area will be blank (`[ ]`):  
  ![Cell run status blank](./assets/readme/jupyterlab-cell-run-indicator-not-run.png)
- If the cell is currently running there will be an astrix in the area (`[*]`):  
  ![Cell run status running](./assets/readme/jupyterlab-cell-run-indicator-running.png)
- If the cell is done running there will be a number in this area (`[8]`), the number is not important:  
  ![Cell run status run](./assets/readme/jupyterlab-cell-run-indicator-run.png)

If a cell is currently runnning and you want to stop it:

1. Select the cell (Just like you would to run the cell)
2. Click the stop button in the top of the site:
