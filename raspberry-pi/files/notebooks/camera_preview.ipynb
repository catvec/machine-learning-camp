Now that we have a `Picamera2` instance we can configure it. We will our camera to show a preview image. 

- To do this we will call the `create_preview_configuration` function
  - It takes a dictionary with the keys `size` and `format`
  - Returns a configuration value
- Then we will pass this configuration to the `configure` function to configure our camera

_Run the cell below_

```python
# RUN ME
config = cam.create_preview_configuration(main={
    "size": (224, 224),
    "format": "BGR888",
}, display="main")
cam.configure(config)
```

Now that we have configured our camera we can tell our camera to open the preview and start capturing images.

- The `start_preview` function creates the preview image
- The `start` function starts capturing images

_Run the cell below_

```python
# RUN ME
cam.start_preview(Preview.QT)
cam.start()
```

If all goes well you should now see your camera on the screen on your Raspberry Pi!

Once you are done viewing your camera run the code below to stop the preview and stop the camera:

```python
# RUN ME
cam.stop_preview()
cam.stop()
```
